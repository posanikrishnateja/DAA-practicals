```python
import time
import random



# 1. BUBBLE SORT

def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True


        if not swapped:
            break

    return arr



# 2. INSERTION SORT

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        # Move elements greater than key one position ahead
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr



# 3. SELECTION SORT

def selection_sort(arr):
    n = len(arr)

    for i in range(n):
        min_index = i

        # Find the smallest element
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        # Swap minimum element with first unsorted element
        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr



# 4. MERGE SORT

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Add remaining elements
    result.extend(left[i:])
    result.extend(right[j:])

    return result



# 5. QUICK SORT

def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)



# EXECUTION TIME TEST

# Generate random data
data = [random.randint(1, 10000) for _ in range(1000)]

# Dictionary containing all sorting algorithms
sorting_algorithms = {
    "Bubble Sort": bubble_sort,
    "Insertion Sort": insertion_sort,
    "Selection Sort": selection_sort,
    "Merge Sort": merge_sort,
    "Quick Sort": quick_sort
}


print("Original array (first 20 elements):")
print(data[:20])
print("\nExecution Time:\n")


for name, algorithm in sorting_algorithms.items():

    # Make a copy so every algorithm gets the same input
    arr = data.copy()

    start_time = time.perf_counter()

    sorted_array = algorithm(arr)

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    print(f"{name:<20} : {execution_time:.8f} seconds")

    # Check whether sorting was successful
    print("Sorted correctly     :", sorted_array == sorted(data))
    print()
```
